
15.3 Show that if $ X_1,...,X_n$ are i.i.d (độc lập cùng phân phối), then 

$$
\varphi_{S_n}(u)= (\varphi_X(u))^n 
$$

where $ S_n = \sum_{j=1}^n X_j $

Ta có : Do các biến ngẫu nhiên  $ X_1,X_2...X_n $ độc lập nên :
        
$$\varphi_{S_n}(u) = \varphi_{X_1}(u).\varphi_{X_2}(u)...\varphi_{X_n}(u)  $$
Do các biến ngẫu nhiên  $ X_1,X_2...X_n $ cùng phân phối , nên 
$$\varphi_{S_n}(u) = \varphi_{X_1}(u).\varphi_{X_1}(u)...\varphi_{X_1}(u)  = (\varphi_X(u))^n  $$


15.14 
Show that $|1- e^{ix}|^2 = 2(1-\cos(x)) \le x^2 $ for all $x \in R$ . Use this to show that $|1 − \varphi_X(u)| \le E\{|uX|\}\\ $

Ta có : 
$
|1- e^{ix}|^2 = |1 - \cos(x) -i\sin(x)|^2 = (1-\cos(x))^2 + (\sin(x))^2 = 1 - 2\cos(x) + \cos^2(x) + \sin^2(x) = 2 - 2\cos(x) = 2(1-\cos(x)) \\
$
Mà $|sin(x)| \le |x| \implies |1 - cos(x)|  = 2 \sin^2(\frac{x}{2} ) \le 2(\frac{x}{2})^2 = \frac{x^2}{2} \Leftrightarrow 2|1-\cos(x)| \le x^2\\
\implies |1- e^{ix}|^2 \le x^2 \implies|1- e^{ix}| \le |x|
$
Ta có :
$|1 − \varphi_X(u)| = |1 - E\{e^{iuX}\}| \le E|(1 - \{e^{iuX}\})|  \le E|(uX)|$

Công thức : 
$|EX| \le E|X|$






15.16 If $\varphi$ is a characteristic function, show that $|\varphi|^2$ is one too. (Hint: Let $X, Y$ be i.i.d. and consider $Z = X − Y $.)

Ta có :

$\varphi_{X-Y} = \varphi_X . \varphi_{(-Y)} =  E\{e^{iuX}\} . E\{e^{-iuY}\} =  E\{e^{iuX}\} . E\{e^{-iuX}\} = E\{e^{iuX}\} . E\{e^{-iuX}\} = |E\{e^{iuX}\}|^2 = |\varphi_X|^2 $


Vậy $|\varphi_X|^2 $ là hàm đặc trưng phân phối $(X-Y)$
